```python
#import libraries
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import random_split, DataLoader
import copy
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as colors
from sklearn.metrics import confusion_matrix, roc_auc_score, RocCurveDisplay
import seaborn as sns

#create functions
def norm(loader):
    """
    This functions takes a data loader and returns the mean and standard deviation of the data set for normalisation purposes
    """
    mean, std, total = 0, 0, 0
    for images,_ in loader:  #calculates the mean and standard deviation of the scans for normalisation
        batch_samples = images.size(0) #takes the size of eah image
        scans = images.view(batch_samples, images.size(1), -1) #flattens each image
        mean += scans.mean(2).sum(0) #calculates mean and standard deviation of each batch
        std += scans.std(2).sum(0)
        total += batch_samples
    mean /= total #calculates mean and standard deviation of the whole data set
    std /= total
    return mean, std

def plot_image(input, target): 
    """
    This function takes 2 parameters:
    input - a tensor of an RGB image to be displayed
    target - a tensor containing the class of the image
    and displays the input tensor as an RGB images with its class
    """
    print('Tumour type', target) #prints the tumour type
    frame = plt.gca()
    frame.axes.get_xaxis().set_visible(False)
    frame.axes.get_yaxis().set_visible(False)
    plt.imshow(input) #displays the image
    plt.show()

def make_loader(dataset, batch_size=32, shuffle=False):
    """
    This function takes a dataset and returns a function that will load the images in batches
    """
    return DataLoader(dataset, batch_size=batch_size, shuffle=shuffle)
    
def get_labels(model):
    """
    This function takes a class which defines the fitting of a model and returns the actual tumour class, predicted tumour class, and probabilities of each prediction for every element in the test data set
    """
    model.eval() #puts the model into evaluationg mode so training stops and we can run the model on the test set 
    ys = [] #creating empty arrays to store values                                    ,
    y_hats = []
    probabilities = []
    with torch.no_grad(): #stops gradient calculation to speed up process
        for batch in test_loader:
            x, y = batch #sets x as the image data and y as the correct tumour type
            x = x.to(device) #ensures images are processed by gpu if possible
            logits = model(x) #stores the predicted tumour type and the confidence of the prediction of each brain scan in the test data set
            probability = torch.softmax(logits, dim=1) #normalises the confidence of the predictions
            predictions = torch.argmax(logits, dim=1).detach().cpu().numpy() #stores the predicted tumour as the class with the highest probability 
            probabilities.extend(probability.detach().cpu().numpy().tolist()) #adds the probabilities from each batch to the list of all probabilities for the test set 
            ys.extend(y.cpu().numpy().tolist()) #adds the actual tumour types of the brain scans in each batch to a list of actual tumour types for the whole test set
            y_hats.extend(predictions.tolist()) #adds the predicted tumour types of the brain scans in each batch to a list of predicted tumour types for the whole test set  
    return ys, y_hats, probabilities

def binarise(target,labels):
    """
    This function takes 
    target: an integer representing which class of tumour we are interested in
    labels: a list of the class of every tumour in the test set
    and returns a numpy array of 1s and 0s where 1s represent the class we are interested in and 0 represents everything else
    """
    ys_bin = np.zeros(len(labels)) #creates an array of 0s to store the binarised values
    for i in range (len(labels)):
        if labels[i] == target:
            ys_bin[i] = 1 #sets all indices of the target tumour class to 1
    return ys_bin
    
def plot_loss(models,result,train_loss,val_loss,titles,show_title):
    """
    This function plots the training and validation loss at each epoch for any model.
    It takes:
    models - an integer representing the amount of models that should have their training and validation loss plotted
    result - a 2D numpy array of every integer from 0 to the amount of epochs used in training the model for every model that should be plotted
    train_loss - a 2D array of the average loss of all batches of training data for each epoch for every model that should be plotted
    val_loss - a 2D array of the average loss of all batches of validation data for each epoch for every model that should be plotted
    titles - a list of strings describing each model that should be plotted
    show_title - a boolean, if True the plot will be shown with a title
    """
    plt.figure() #plotting a logarithmic graoh of train and validation loss against epoch for each model
    title = 'Loss of train and validation sets across all epochs \n '
    for i in range (models): #plots the desired amount of models' losses
        train_label, val_label = 'Train - '+titles[i],'Validation - '+titles[i] #creates legend labels
        if i!=(models-1): #creates the title of the graph
            title += titles[i]+' and '
        else:
            title += titles[i]
        plt.plot(result[i], train_loss[i], label=train_label) #plots train loss for each model
        plt.plot(result[i], val_loss[i], label=val_label) #plots validation loss for each model
    plt.autoscale(enable=True, axis='both', tight=True) #adjusts the axis sizes for each dataset
    plt.yscale('log') #logarithmic scale
    plt.grid(visible=True, which='both') #uses grid lines for easier comparison between plots
    plt.legend() #displays legend
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    if show_title: #displays the title if desired
        plt.title(title)
    plt.show()
    
def confusion_matrices(model,title,show_title=True):
    """
    This function produces a confusion matrix for a model.
    It takes 3 parameters:
    model - the class containing the fitting algorithm of a model
    title - a string describing the model
    show_title - a boolean, if True the plot will be shown with a title
    """
    ys, y_hats, probabilities = get_labels(model) #calls the get_labels function to find the predicted and actual tumour types of each image
    cm = confusion_matrix(ys,y_hats) #returns a confusion matrix for eaach model using scikit learn
    cm_percentages = cm/cm.sum(axis=1)[:,None] *100 #turns the confusion matrix values into percentaage for ease of comparison on the plot
    plt.figure()
    sns.heatmap(cm_percentages, annot=True,cmap='BuPu', xticklabels=['Meningioma', 'Glioma', 'Pituitary'], yticklabels=['Meningioma', 'Glioma', 'Pituitary']) #uses seaborn to plot the confusion matrix with a colour scale
    plt.ylabel('True tumour type')
    plt.xlabel('Predicted tumour type')
    if show_title: #shows graph title if desired
        title = 'Confusion matrix '+title
        plt.title(title)
    plt.show()
    return cm

def tpr(model,title):
    """
    This function takes two parameters:
    model - the class containing the fitting algorithm of a model
    title - a string describing the model
    and returns the true positive rate and false positive rate of the model
    """
    conf_matrix = confusion_matrices(model,title) #calls the function to return the confusion matrix of the model
    num_predictions = np.sum(conf_matrix) #sums how many images the model has predicted tumour types for in the test set
    num_true_pred = np.trace(conf_matrix) #sums the diagonal elements of the confusion matrix which are the ammount of images that were assigned the correct tumour type (the amount of true positives)
    true_pos_rate = num_true_pred/num_predictions #calculates the fraction of predictions that were correct
    false_pos_rate = 1 - true_pos_rate #calculates and prints the false positive rate by calculating what fraction of predictions weren't correct
    return true_pos_rate, false_pos_rate
    
def roc(target,model):
    """
    This function takes on parameter:
    target - an integer representing which class of tumour we are interested in
    model - the class containing the fitting algorithm of a model
    and displays the ROC curve produced by the model for the tumour type we are interested in, and returns its AUC
    """
    ys, y_hats, probabilities = get_labels(model) #calls the get_labels function to find the predicted and actual tumour types of each image
    ys_bin = binarise(target,labels=ys) #calls the binarise function for plotting the ROC
    probabilities = np.array(probabilities)
    roc_curve = RocCurveDisplay.from_predictions(ys_bin, probabilities[:,target], name = 'ROC', plot_chance_level=True) #plots the ROC for a tumour class using scikit learn    
    auc_score = roc_auc_score(ys_bin, probabilities[:, target]) #calculates and AUC of ROC curve using scikit learn
    return roc_curve, auc_score
     
```

```python
#convolutional neural network
class brain_scan_CNN(nn.Module):
    def __init__(self): #setting up the network architecture
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels, 32, kernel_size=3, padding=1) # first convolutional layer
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1) #second convolutional layer
        self.conv3 = nn.Conv2d(64,128, kernel_size=3, padding=1) #third convolutional layer added as with 2 convolution layers the predictions were poor on the first run and better on the second
        self.bn1 = nn.BatchNorm2d(32) #normalises input after each convoliutional layer
        self.bn2 = nn.BatchNorm2d(64)
        self.bn3 = nn.BatchNorm2d(128)
        self.pool = nn.MaxPool2d(2, 2) #max pooling
        self.relu = nn.ReLU() #rectified linear activation function
        self.drop_out = nn.Dropout(p=0.2) #randomly sets 20% of tensors to 0 to reduce overfitting
        self.fc1 = nn.Linear(128 * 16 * 16, 128) #fully connected layers (layers which connect to every neuron of the prior layer)
        self.fc2 = nn.Linear(128, 32)
        self.fc3 = nn.Linear(32, in_channels)
        self.loss = nn.CrossEntropyLoss() #softmax included in cross entropy loss and so softmax is not used here
        
    def forward(self, x): #defines forward pass of the network
        x = self.pool(self.relu(self.bn1(self.conv1(x))))
        x = self.pool(self.relu(self.bn2(self.conv2(x))))
        x = self.pool(self.relu(self.bn3(self.conv3(x))))
        x = x.view(x.size(0), -1) #allows a flatten version of the scans to be viewed
        out = self.relu(self.fc1(x))
        out = self.relu(self.fc2(out))
        out = self.drop_out(out)
        out = self.fc3(out)
        return out

    def training_step(self, batch, device=None): #function to process a batch of the training data set through the forward pass
        x, y = batch #takes a batch of data and sets the image tensors to x and their tumour type to y
        if device is not None: #makes sure the data is processed on the best device available
            x = x.to(device)
            y = y.to(device)
        y_hat = self(x) #finds the models predicted tumour classification
        loss = self.loss(y_hat, y) #calculates cross entropy loss
        return loss

class CNN_Trainer:
    def __init__(self, model, n_epochs, lr, train_loader, validation_loader, device, patience, min_delta): #initialises the class
        self.model = model
        self.n_epochs = n_epochs
        self.loss_train = np.zeros(n_epochs)
        self.loss_val = np.zeros(n_epochs)
        self.train_loader = train_loader
        self.validation_loader = validation_loader
        self.device = device
        self.optimizer = torch.optim.Adam(model.parameters(), lr=(lr)) #stochastic optimisation
        lr_func = lambda x:0.925 #sets the function for the learning rate scheduler, this means for an epoch n the learning rate is initial learning rate * 0.925^n
        self.scheduler = torch.optim.lr_scheduler.MultiplicativeLR(optimizer=self.optimizer,lr_lambda=lr_func) #uses multiplicative learning rate scheduler
        self.model.to(self.device) #ensures the model runs on the best available device
        self.patience = patience 
        self.min_delta = min_delta
        self.best_state = None
        self.best_val = float('inf')
        self.bad_epochs = 0
     
    def fit(self):
        for epoch in range(self.n_epochs):
            self.model.train() #puts model in training mode
            loss_t = []
            for batch in self.train_loader:
                loss = self.model.training_step(batch, device=self.device) #calculates cross entropy loss of each batch of training data from the training step function defined above
                self.optimizer.zero_grad() #sets gradients to zero so weights and biases can be correctly updated in back propagation
                loss.backward() #backpropagation
                self.optimizer.step() #perform an optimisation step
                loss_t.append(loss.item())
            self.loss_train[epoch] = np.mean(loss_t) #reports the training loss as the average loss from all training batches
            self.scheduler.step() #alters learning rate each epoch

            self.model.eval() #puts model in evaluation mode
            loss_v = []
            with torch.no_grad(): #disables gradient calculation
                for batch in self.validation_loader:
                    loss = self.model.training_step(batch, device=self.device).item() #calculates cross entropy loss of each batch of the validation data set using the training step function
                    loss_v.append(loss)
            self.loss_val[epoch] = np.mean(loss_v) #reports the validation loss as the average loss of all validation batches

            print('Epoch',(epoch+1),'/',self.n_epochs,'train_loss:',self.loss_train[epoch],'validation_loss:',self.loss_val[epoch]) #prints the training and validation loss from each epoch

            if self.loss_val[epoch] < (self.best_val - self.min_delta): #if the validation loss of this epoch is the lowest then it is stored as the new lowest loss and the amount of plateaud epochs is set to 0
                self.best_val = self.loss_val[epoch] 
                self.best_state = copy.deepcopy(self.model.state_dict())
                self.bad_epochs = 0
            else: #counts how many epochs in a row have not given a new lowest validation loss, if this is greater than the patience then the model is assumed to have plateaud and stops early to reduce overfitting
                self.bad_epochs += 1
                if self.bad_epochs >= self.patience:
                    print('Early stopping triggered at epoch',epoch+1,'best validation_loss:',self.best_val)
                    break
        
        if self.best_state is not None: #restores best weights so the model with the lowest validation loss is returned
            self.model.load_state_dict(self.best_state)
```

```python
#residual neural network
class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1): #setting up architecture of the residual block
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels, out_channels, 3, stride=stride, padding=1) #first convolution layer
        self.bn1 = nn.BatchNorm2d(out_channels) #normalises input
        self.activation = nn.ReLU() #uses rectified linear activation function
        self.conv2 = nn.Conv2d(out_channels, out_channels, 3, padding=1) #second convolution layer
        self.bn2 = nn.BatchNorm2d(out_channels) #normalises
        if stride != 1 or in_channels != out_channels: #if the size or number of channels of the input scan tensor changes then the dimensions of the shortcut change to match
            self.shortcut = nn.Sequential(nn.Conv2d(in_channels, out_channels, 1, stride=stride),nn.BatchNorm2d(out_channels))
        else: #otherwise the shortcut stays as it was and does not change the input
            self.shortcut = nn.Identity()

    def forward(self, x): #sets up forward pass of the residual block
        identity = self.shortcut(x) #returns the shortcut 
        out = self.activation(self.bn1(self.conv1(x))) 
        out = self.bn2(self.conv2(out))
        out = self.activation(out + identity) # adds the shortcut to the output and applies the activation
        return out
        
class ResNetClassifier(nn.Module):
    def __init__(self, in_channels=3): #setting up resnet architecture
        super().__init__()
        self.conv = nn.Conv2d(in_channels, 32, kernel_size=3, padding=1) #initial convolutional layer
        self.bn = nn.BatchNorm2d(32) #normalisation
        self.layer1 = nn.Sequential(ResidualBlock(32, 32),ResidualBlock(32, 32)) #first residual layer
        self.layer2 = nn.Sequential(ResidualBlock(32, 64, stride=2), ResidualBlock(64,64)) #second residual layer
        self.layer3 = nn.Sequential(ResidualBlock(64, 128, stride=2), ResidualBlock(128,128)) #third residual 
        self.pool = nn.MaxPool2d(32, 32) #max pooling
        self.drop_out = nn.Dropout(p=0.2) #randomly sets 20% of inputs to 0 to reduce overfitting
        self.fc = nn.Linear(128, 3) #fully connected layer
        self.loss = nn.CrossEntropyLoss() #loss function - includes softmax

    def forward(self, x): #defines forward pass of the network
        x = torch.relu(self.bn(self.conv(x))) #rectified linear activation function
        x = self.layer1(x) #residual layers and pooling
        x = self.layer2(x)
        x = self.pool(self.layer3(x))
        x = self.drop_out(x)
        x = x.view(x.size(0),-1) #allows a flatten version of the output to be viewed for the fully connected layer
        x = self.fc(x)
        return x

    def training_step(self, batch, device=None): #function to process a batch of the training data set through the forward pass
        x, y = batch #takes a batch of data and sets the image tensors to x and their tumour type to y
        if device is not None: #makes sure the data is processed on the best device available
            x = x.to(device)
            y = y.to(device)
        y_hat = self(x) #finds the models predicted tumour classification
        loss = self.loss(y_hat, y) #calculates cross entropy loss
        return loss
        
class ResNet_Trainer:
    def __init__(self, model, n_epochs, lr, train_loader, validation_loader, device, patience=3, min_delta=0.0): #initialises the class
        self.model = model
        self.n_epochs = n_epochs
        self.loss_train = np.zeros(n_epochs)
        self.loss_val = np.zeros(n_epochs)
        self.train_loader = train_loader
        self.validation_loader = validation_loader
        self.device = device
        self.optimizer = torch.optim.Adam(model.parameters(), lr=(lr)) #stochastic optimisation
        lr_func = lambda x:0.925  #sets the function for the learning rate scheduler so that for an epoch n the learning rate is initial learning rate * 0.96^n
        self.scheduler = torch.optim.lr_scheduler.MultiplicativeLR(optimizer=self.optimizer, lr_lambda=lr_func) #uses multiplicative learning rate scheduler
        self.model.to(self.device)
        self.patience = patience
        self.min_delta = min_delta
        self.best_state = None
        self.best_val = float('inf')
        self.bad_epochs = 0

    def fit(self):
        for epoch in range(self.n_epochs):
            self.model.train() #puts the model into training mode
            loss_t = []
            for batch in self.train_loader:
                loss = self.model.training_step(batch, device=self.device) #calculates loss of each batch of training data using the training step function defined above
                self.optimizer.zero_grad() #sets gradients to zero so weights and biases can be correctly updated in back propagation
                loss.backward() #backpropagation
                self.optimizer.step() #perform an optimisation step
                loss_t.append(loss.item())
            self.loss_train[epoch] = np.mean(loss_t) #reports the training loss as the average loss from all training batches
            self.scheduler.step() #alters learning rate each epoch

            self.model.eval() #puts the model into evaluation mode and out of training mode
            loss_v = []
            with torch.no_grad(): #disables gradient calculation
                for batch in self.validation_loader:
                    loss = self.model.training_step(batch, device=self.device).item() #calculates cross entropy loss of each batch of the validation data set using the training step function
                    loss_v.append(loss)
            self.loss_val[epoch] = np.mean(loss_v) #reports the validation loss as the average loss of all validation batches

            print('Epoch',(epoch+1),'/',self.n_epochs,'train_loss:',self.loss_train[epoch],'validation_loss:',self.loss_val[epoch]) #prints the training and validation loss from each epoch

            if self.loss_val[epoch] < (self.best_val - self.min_delta): #if the validation loss of this epoch is the lowest then it is stored as the new lowest loss and the amount of plateaud epochs is set to 0
                self.best_val = self.loss_val[epoch]
                self.best_state = copy.deepcopy(self.model.state_dict())
                self.bad_epochs = 0
            else: #counts how many epochs in a row have not given a new lowest validation loss, if this is greater than the patience then the model is assumed to have plateaud and stops early to reduce overfitting
                self.bad_epochs += 1
                if self.bad_epochs >= self.patience:
                    print('Early stopping triggered at epoch',(epoch+1),'best validation_loss:',(self.best_val))
                    break

        if self.best_state is not None: #restores best weights so the model with the lowest validation loss is returned
            self.model.load_state_dict(self.best_state)
```
